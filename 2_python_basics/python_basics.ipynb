## 1. Типи даних

```python
a = "змінна з текстом"
b = 1
b1 = 1.1
c = ["a", 1, 1.25, "Слово", a]
d = {"a": "Слово", "b": 1, a: b}
e = ("a", a)
f = {"ss", a + str(b)}
print(type(a).__name__, type(b).__name__, type(b1).__name__)
print(c)
print(d)
print(e)
print(f)
```

## 2. Константи, keywords і вбудовані функції

```python
import keyword
print("True:", True)
print("None:", None)
print("NotImplemented:", NotImplemented)
print("Перші reserved keywords:", ", ".join(keyword.kwlist[:15]))
print(abs(-12.5), abs(-12.5) == abs(12.5))
print(len([1, 2, 3]), repr("text"))
```

## 3. Цикли та розгалуження

```python
from random import randint
letters = ["a", "b", "c"]
for index, letter in enumerate(letters):
    print(f"На позиції {index} знаходиться буква {letter}")
value = randint(0, 1)
print("Випало 1: умова виконана." if value else "Випало 0: умова не виконана.")
```

## 4. Обробка помилок і context manager

```python
try:
    print(10 / 0)
except ZeroDivisionError as error:
    print("Перехоплена помилка:", error)
finally:
    print("Блок finally виконався.")

with open("python_basics.py", "r", encoding="utf-8") as source_file:
    print("Перша строка:", source_file.readline().strip())
```

## 5. Лямбда-функції

Лямбда-функція — це короткий спосіб описати невелику анонімну функцію без `def`. У прикладі вона підносить кожне число до квадрата через `map`.

```python
square = lambda value: value ** 2
numbers = [1, 2, 3, 4]
print(list(map(square, numbers)))
```
