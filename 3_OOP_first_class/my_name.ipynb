# Перший клас в Python

У цій роботі продемонстровано створення класу, атрибутів об'єкта, властивостей, методів і класових змінних.

```python
class MyName:
    total_names = 0

    def __init__(self, name=None, domain="itcollege.lviv.ua"):
        if name is None:
            name = "Anonymous"
        elif not isinstance(name, str) or not name.isalpha():
            raise ValueError("Ім'я може містити лише літери!")

        self.name = name[0].upper() + name[1:]
        self.domain = domain
        MyName.total_names += 1
        self.my_id = MyName.total_names

    @property
    def whoami(self):
        return f"My name is {self.name}"

    def create_email(self, domain=None):
        email_domain = domain if domain is not None else self.domain
        return f"{self.name.lower()}@{email_domain}"

    @property
    def my_email(self):
        return self.create_email()

    @property
    def name_length(self):
        return len(self.name)

    @property
    def full_name(self):
        return f"User #{self.my_id}: {self.name} ({self.my_email})"

    @classmethod
    def anonymous_user(cls):
        return cls("Anonymous")

    @staticmethod
    def say_hello(message="Hello to everyone!"):
        return f"You say: {message}"
```

## Створення об'єктів

```python
names = ("Vitaliy", "Bohdan", "Marta", None)
all_names = {name: MyName(name) for name in names}

for name, user in all_names.items():
    print(user.whoami)
    print(user.full_name)
    print("Довжина імені:", user.name_length)
    print("Привітання:", user.say_hello("Welcome to Python OOP!"))
    print("Email з новим доменом:", user.create_email("lpnu.ua"))
    print("-" * 30)

print("Кількість елементів у names:", len(names))
print("Кількість створених об'єктів:", MyName.total_names)
```

## Перевірка некоректного імені

```python
try:
    MyName("Vitaliy123")
except ValueError as error:
    print(type(error).__name__ + ":", error)
```

## Збереження у файл

```python
def save_to_file(user, filename="users.txt"):
    with open(filename, "a", encoding="utf-8") as file:
        file.write(user.full_name + "\n")

save_to_file(next(iter(all_names.values())))
```
